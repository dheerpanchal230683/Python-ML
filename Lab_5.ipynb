[1,4,9,16,25] → These are your y-values (squares of x-values).

'r^-' → This is a format string that styles the plot:

'r' → Red color

'^' → Triangle-up markers for each point

'-' → Solid line connecting the points

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data for a box plot
data_group1 = np.random.normal(100, 10, 200) # Mean 100, Std Dev 10, 200 points
data_group2 = np.random.normal(90, 20, 200)  # Mean 90, Std Dev 20, 200 points
data_group3 = np.random.normal(110, 5, 200)  # Mean 110, Std Dev 5, 200 points

plot_data = [data_group1, data_group2, data_group3]
labels = ['Group 1', 'Group 2', 'Group 3']

# Create a figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Create the box plot
ax.boxplot(plot_data, labels=labels, patch_artist=True, medianprops={'color': 'red'})

# Set labels and title
ax.set_xlabel('Data Groups')
ax.set_ylabel('Values')
ax.set_title('Box Plot of Sample Data')

# Add grid for better readability
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Sample data for multiple groups
np.random.seed(42)
data_group1 = np.random.normal(100, 10, 100)
data_group2 = np.random.normal(90, 15, 100)
data_group3 = np.random.normal(110, 8, 100)
data_group4 = np.random.normal(95, 12, 100)

all_data = [data_group1, data_group2, data_group3, data_group4]
labels = ['Category A', 'Category B', 'Category C', 'Category D']

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Create a 3D box plot by iterating through categories
for i, data in enumerate(all_data):
    # Calculate box plot statistics (quartiles, whiskers, outliers)
    q1, median, q3 = np.percentile(data, [25, 50, 75])
    iqr = q3 - q1
    whisker_low = np.max([data.min(), q1 - 1.5 * iqr])
    whisker_high = np.min([data.max(), q3 + 1.5 * iqr])

    # Plot median
    ax.plot([i+1, i+1], [median, median], [0, 0], marker='_', markersize=10, color='red', zorder=10)

    # Plot box (IQR)
    ax.bar3d(i+1 - 0.2, q1, 0, 0.4, iqr, 0, color='cyan', alpha=0.6)
    ax.bar3d(i+1 - 0.2, median, 0, 0.4, 0, 0, color='red', zorder=11) # Median line

    # Plot whiskers
    ax.plot([i+1, i+1], [whisker_low, q1], [0, 0], color='blue', linestyle='-', linewidth=2)
    ax.plot([i+1, i+1], [q3, whisker_high], [0, 0], color='blue', linestyle='-', linewidth=2)
    ax.plot([i+1 - 0.1, i+1 + 0.1], [whisker_low, whisker_low], [0, 0], color='blue', linewidth=2)
    ax.plot([i+1 - 0.1, i+1 + 0.1], [whisker_high, whisker_high], [0, 0], color='blue', linewidth=2)

    # Plot outliers (optional)
    outliers = data[(data < whisker_low) | (data > whisker_high)]
    if len(outliers) > 0:
        ax.scatter([i+1] * len(outliers), outliers, [0]*len(outliers), color='red', marker='x', s=50, zorder=12)

# Set labels and title
ax.set_xlabel('Category')
ax.set_ylabel('Value')
ax.set_zlabel('') # No specific Z-label for this representation
ax.set_title('Simulated 3D Box Plot for Multiple Categories')

# Set x-axis ticks to be the category labels
ax.set_xticks(np.arange(1, len(labels) + 1))
ax.set_xticklabels(labels)

# Adjust viewpoint
ax.view_init(elev=20, azim=-35)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Create a figure and a 3D axes
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Define the x and y ranges for the surface
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)

# Define a function to plot (e.g., a 2D sine wave)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Add labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis (sin(R))')
ax.set_title('3D Surface Plot of sin(sqrt(X^2 + Y^2))')

# Add a color bar which maps values to colors.
fig.colorbar(surf, shrink=0.5, aspect=5)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data from the existing pie chart in the notebook
labels = ['Class 0', 'Class 1']
sizes = [4, 6]  # Example counts

# 'explode' a slice if desired, this gives a 3D effect
# For example, explode the first slice slightly
explode = (0.1, 0) # 0.1 for the first slice, 0 for the second

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
       shadow=True, startangle=90, colors=['lightcoral', 'lightskyblue'])

# Equal aspect ratio ensures that pie is drawn as a circle.
ax.axis('equal')
ax.set_title('Simulated 3D Pie Chart (with Explode and Shadow)')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure and a 2x2 grid of subplots
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
fig.suptitle('Four Different Plots in Subplots', fontsize=16)

# Plot 1: Line Plot
x1 = np.linspace(0, 10, 100)
y1 = np.sin(x1)
axes[0, 0].plot(x1, y1, color='blue')
axes[0, 0].set_title('Line Plot (sin(x))')
axes[0, 0].set_xlabel('X-axis')
axes[0, 0].set_ylabel('Y-axis')
axes[0, 0].grid(True)

# Plot 2: Scatter Plot
x2 = np.random.rand(50) * 10
y2 = np.random.rand(50) * 10
axes[0, 1].scatter(x2, y2, color='red', marker='o')
axes[0, 1].set_title('Scatter Plot')
axes[0, 1].set_xlabel('X-axis')
axes[0, 1].set_ylabel('Y-axis')
axes[0, 1].grid(True)

# Plot 3: Bar Chart
categories = ['A', 'B', 'C', 'D']
values = [23, 56, 12, 45]
axes[1, 0].bar(categories, values, color=['green', 'yellow', 'orange', 'purple'])
axes[1, 0].set_title('Bar Chart')
axes[1, 0].set_xlabel('Category')
axes[1, 0].set_ylabel('Value')
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.7)

# Plot 4: Histogram
data_hist = np.random.randn(1000)
axes[1, 1].hist(data_hist, bins=30, color='cyan', edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Histogram')
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(True)

# Adjust layout to prevent overlapping titles/labels
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust rect to make space for suptitle
plt.show()
```
